```python
cursor.execute("SELECT * FROM categories")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
cursor.execute("SELECT * FROM competitions")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#1 List all competitions along with their category name
cursor.execute("SELECT category_name  from categories")
 

out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#2 Count the number of competitions in each category
cursor.execute("SELECT categories.category_name, COUNT(competitions.competition_id) AS competition_count FROM categories LEFT JOIN competitions ON categories.category_id = competitions.category_id GROUP BY categories.category_name")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#3 Find all competitions of type 'doubles'
cursor.execute("select * from competitions WHERE type='doubles'")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#4 Get competitions that belong to a specific category (e.g., ITF Men)
cursor.execute("select * from competitions WHERE gender= 'Men'")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#5 Identify parent competitions and their sub-competitions
cursor.execute("""
    SELECT 
    parent.competition_id AS parent_competition_id,
    parent.competition_name AS parent_competition_name,
    sub.competition_id AS sub_competition_id,
    sub.competition_name AS sub_competition_name
FROM competitions parent
JOIN competitions sub
    ON parent.competition_id = sub.parent_id
ORDER BY parent.competition_name, sub.competition_name;
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#6 Analyze the distribution of competition types by category
cursor.execute("""
SELECT 
    cat.category_name,  
    c.type, 
    COUNT(*) AS competition_count 
FROM competitions c 
JOIN categories cat 
    ON c.category_id = cat.category_id 
GROUP BY cat.category_name, c.type 
ORDER BY cat.category_name, competition_count DESC
""")


out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#7 List all competitions with no parent (top-level competitions)
cursor.execute("SELECT * FROM competitions WHERE  NOT parent_id IS NOT NULL")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
cursor.execute("SELECT * FROM complexes")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
cursor.execute("SELECT * FROM venues")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#1 List all venues along with their associated complex name
cursor.execute("""SELECT 
    v.venue_id,
    v.venue_name,
    v.city_name,
    v.country_name,
    v.country_code,
    v.timezone,
    c.complex_name
FROM venues v
LEFT JOIN complexes c 
    ON v.complex_id = c.complex_id
ORDER BY v.venue_name;
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#2 Count the number of venues in each complex
cursor.execute("select complex_id,count(venue_name)AS venue_count FROM venues GROUP BY complex_id")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#3 Get details of venues in a specific country (e.g., Chile)
cursor.execute("select * FROM venues WHERE country_name = 'chile'")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#4 Identify all venues and their timezones
cursor.execute("select venue_id,venue_name,timezone FROM venues")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#5Find complexes that have more than one venue
cursor.execute("""
SELECT 
    c.complex_id,
    c.complex_name,
    COUNT(v.venue_id) AS venue_count
FROM complexes c
JOIN venues v 
    ON c.complex_id = v.complex_id
GROUP BY c.complex_id, c.complex_name
HAVING COUNT(v.venue_id) > 1
ORDER BY venue_count DESC
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#6List venues grouped by country
cursor.execute("select country_name,venue_name,COUNT(venue_name)FROM venues GROUP BY country_name")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#7 Find all venues for a specific complex (e.g., Nacional)
cursor.execute("""SELECT 
    v.venue_id,
    v.venue_name,
    v.city_name,
    v.country_name,
    v.timezone,
    c.complex_name
FROM venues v
JOIN complexes c 
    ON v.complex_id = c.complex_id
WHERE c.complex_name = 'Nacional'
ORDER BY v.venue_name;
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
cursor.execute("SELECT * FROM competitor")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
cursor.execute("SELECT * FROM ranking")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#1 Get all competitors with their rank and points.
cursor.execute("""SELECT 
    c.competitor_id, 
    c.name, 
    r.rank, 
    r.points
FROM competitor c
JOIN ranking r 
    ON c.competitor_id = r.competitor_id
ORDER BY r.rank;
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#2 Find competitors ranked in the top 5
cursor.execute("""SELECT 
    c.competitor_id, 
    c.name, 
    r.rank, 
    r.points
FROM competitor c
JOIN ranking r 
    ON c.competitor_id = r.competitor_id
ORDER BY r.rank
LIMIT 5""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#3 List competitors with no rank movement (stable rank)
cursor.execute("""
    SELECT 
    c.competitor_id, 
    c.name, 
    r.rank, 
    r.points, 
    r.movement
FROM competitor c
JOIN ranking r 
    ON c.competitor_id = r.competitor_id
WHERE r.movement = 0
ORDER BY r.rank
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#4 Get the total points of competitors from a specific country (e.g., Croatia)
cursor.execute("""
    SELECT 
    c.country,
    SUM(r.points) AS total_points
FROM competitor c
JOIN ranking r 
    ON c.competitor_id = r.competitor_id
WHERE c.country = 'Croatia'
GROUP BY c.country;
""")
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#Count the number of competitors per country
cursor.execute("""
    SELECT 
    c.country, 
    COUNT(c.competitor_id) AS competitor_count
FROM competitor c
GROUP BY c.country
ORDER BY competitor_count DESC;
    """)
out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```

```python
#6Find competitors with the highest points in the current week
cursor.execute("""
    SELECT 
    c.competitor_id, 
    c.name, 
    r.rank, 
    r.points, 
    r.movement
FROM competitor c
JOIN ranking r 
    ON c.competitor_id = r.competitor_id
ORDER BY r.points DESC
LIMIT 1;



""")

out=cursor.fetchall()
print(tabulate(out, headers=[i[0] for i in cursor.description],  tablefmt='psql'))
```
